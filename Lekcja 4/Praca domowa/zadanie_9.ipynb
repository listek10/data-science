```python
print(f"{'Oryginalna':>12} | {'Znormalizowana':>14} | {'W [0, 1]?':>9}")
print("-" * 42)

for value in values:
    normalized = (value - col_min) / (col_max - col_min)
    in_range = 0 <= normalized <= 1
    print(f"{value:>12} | {normalized:>14.4f} | {str(in_range):>9}")
```
