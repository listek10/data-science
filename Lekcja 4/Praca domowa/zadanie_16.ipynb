```python
tp = 45
tn = 38
fp = 7
fn = 10
```

```python
total = tp + tn + fp + fn
accuracy = (tp + tn) / total
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1_score = 2 * precision * recall / (precision + recall)
```

```python
print(f"{'':8}{'Predicted':^16}")
print(f"{'':8}{'0':>7} {'1':>7}")
print(f"{'Actual':<8}{'0':<2}[{tn:^5}] [{fp:^5}]")
print(f"{'':8}{'1':<2}[{fn:^5}] [{tp:^5}]")
print()
print(f"Accuracy:  {accuracy:.2%}")
print(f"Precision: {precision:.2%}")
print(f"Recall:    {recall:.2%}")
print(f"F1-Score:  {f1_score:.2%}")
```
