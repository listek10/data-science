```python
age_raw = " 45.7 "
gender_raw = "male"
bp_raw = "120/80"
height_raw = " 175.5cm "
```

```python
print("RAPORT WALIDACJI")
print("=" * 50)

# Wiek
age_clean = age_raw.strip()
age = float(age_clean)
age_ok = 0 <= age <= 120
print(f"Wiek: '{age_raw}' → {age}")
print(f"  Status: {'OK' if age_ok else 'BŁĄD'} (zakres [0, 120])")

# Płeć
gender = gender_raw.strip().lower()
gender_ok = gender in ("male", "female")
print(f"Płeć: '{gender_raw}' → '{gender}'")
print(f"  Status: {'OK' if gender_ok else 'BŁĄD'} (male/female)")

# Ciśnienie krwi
systolic_str, diastolic_str = bp_raw.strip().split("/")
systolic = int(systolic_str)
diastolic = int(diastolic_str)
bp_ok = (90 <= systolic <= 180) and (60 <= diastolic <= 120)
print(f"Ciśnienie: '{bp_raw}' → skurczowe={systolic}, rozkurczowe={diastolic}")
print(f"  Status: {'OK' if bp_ok else 'BŁĄD'} (skurczowe [90, 180], rozkurczowe [60, 120])")

# Wzrost
height_clean = height_raw.strip().lower().replace("cm", "").strip()
height = float(height_clean)
height_ok = 50 <= height <= 250
print(f"Wzrost: '{height_raw}' → {height} cm")
print(f"  Status: {'OK' if height_ok else 'BŁĄD'} (zakres [50, 250] cm)")

print("=" * 50)
all_ok = age_ok and gender_ok and bp_ok and height_ok
print(f"Wszystkie pola poprawne: {all_ok}")
```
