```python
table = "patients"
columns = ["patient_id", "age", "diagnosis"]
age_threshold = 40
diagnoses = ["diabetes", "hypertension"]
limit = 100
```

```python
columns_sql = ", ".join(columns)
diagnoses_sql = "(" + ", ".join(f"'{d}'" for d in diagnoses) + ")"

query = f"""SELECT {columns_sql}
FROM {table}
WHERE age > {age_threshold} AND diagnosis IN {diagnoses_sql}
ORDER BY age DESC
LIMIT {limit};"""

print(query)
```
