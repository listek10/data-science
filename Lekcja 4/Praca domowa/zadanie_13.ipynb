```python
values = [120, 125, 118, 190, 122]
n = len(values)
```

```python
mean = sum(values) / n
variance = sum((x - mean) ** 2 for x in values) / n
std = variance ** 0.5

print(f"Średnia (wszystkie pomiary): {mean:.2f} mmHg")
print(f"Odchylenie std (wszystkie): {std:.2f} mmHg")
print(f"Próg 3×std: ±{3 * std:.2f} mmHg")
print()
print("Uwaga: przy n=5 klasyczna reguła 3-sigma 'chowa' outliera")
print("(190 ma z≈1.99). Dlatego dla każdego punktu liczymy")
print("średnią i std z POZOSTAŁYCH wartości (leave-one-out).")
print("-" * 40)
```

```python
for i, x in enumerate(values):
    others = [v for j, v in enumerate(values) if j != i]
    mean_others = sum(others) / len(others)
    std_others = (sum((v - mean_others) ** 2 for v in others) / len(others)) ** 0.5
    is_anomaly = abs(x - mean_others) > 3 * std_others

    if is_anomaly:
        print(f"Wartość: {x} - ANOMALIA!")
    else:
        print(f"Wartość: {x} - OK")
```
