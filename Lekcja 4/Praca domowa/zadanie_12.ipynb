```python
n_samples = 50000
batch_size = 128
epochs = 50
time_per_batch = 0.15
```

```python
full_batches = n_samples // batch_size
remainder = n_samples % batch_size
batches_per_epoch = full_batches + (1 if remainder > 0 else 0)

total_batches = batches_per_epoch * epochs
total_seconds = total_batches * time_per_batch

hours = int(total_seconds // 3600)
minutes = int((total_seconds % 3600) // 60)
seconds = int(total_seconds % 60)

print(f"Batchy na epokę: {full_batches} pełnych + reszta {remainder} → {batches_per_epoch}")
print(f"Łącznie batchy: {total_batches}")
print(f"Czas całkowity: {total_seconds:.1f} s")
print(f"Szacowany czas: {hours} h {minutes} min {seconds} s")
```
