```python
temperature_C = 23.5
```

```python
Fahrenheit = (temperature_C * 9/5) + 32
Kelvin = temperature_C + 273.15
```

```python
print (f"Temperatura w stopniach Celsjusza: {temperature_C:.2f} °C")
print (f"Temperatura w Fahrenheitach: {Fahrenheit:.2f} °F")
print (f"Temperatura w Kelwinach: {Kelvin:.2f} K")
```
