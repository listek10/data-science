```python
weights = [65.5, 78.2, 54.8, 92.1, 70.0]
n = len(weights)
```

```python
mean = sum(weights) / n
std = (sum((x - mean) ** 2 for x in weights) / n) ** 0.5

print(f"Średnia: {mean:.2f} kg")
print(f"Odchylenie std: {std:.2f} kg")
print("-" * 45)

z_scores = []
for x in weights:
    z = (x - mean) / std
    z_scores.append(z)

    if z > 2:
        label = "bardzo wysoka"
    elif z < -2:
        label = "bardzo niska"
    elif abs(z) <= 1:
        label = "typowa"
    else:
        label = "umiarkowana"

    print(f"{x:>5} kg → z = {z:+.2f} ({label})")
```

```python
z_mean = sum(z_scores) / n
z_std = (sum((z - z_mean) ** 2 for z in z_scores) / n) ** 0.5

print(f"Średnia z-scores: {z_mean:.10f} (≈ 0)")
print(f"Odchylenie z-scores: {z_std:.10f} (≈ 1)")
```
