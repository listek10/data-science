```python
weight = 80
height = 1.75
```

```python
BMI = weight / (height ** 2)
```

```python
print(f"Twoje BMI wynosi: {BMI:.2f}")
```

```python
if BMI < 18.5:
    print("Masz niedowagę.")
elif 18.5 <= BMI < 25:
    print("Masz prawidłową wagę.")
elif 25 <= BMI < 30:
    print("Masz nadwagę.")
```
