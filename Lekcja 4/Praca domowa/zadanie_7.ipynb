```python
acc_a, time_a = 0.87, 120
acc_b, time_b = 0.89, 450
threshold = 0.02
```

```python
# round() usuwa błąd zmiennoprzecinkowy: 0.89 - 0.87 daje 0.020000000000000018
acc_diff = round(acc_b - acc_a, 4)
time_diff = time_b - time_a

significant = abs(acc_diff) > threshold
b_more_accurate = acc_diff > 0
a_faster = time_a < time_b

print(f"Różnica accuracy (B - A): {acc_diff:+.2%}")
print(f"Różnica znacząca (> {threshold:.0%}): {significant}")
print(f"Różnica czasu treningu (B - A): {time_diff:+} s ({time_b / time_a:.2f}x)")
print(f"Model A szybszy: {a_faster}")
```

```python
if significant and b_more_accurate:
    recommendation = "Model B – znacząco lepsza accuracy uzasadnia dłuższy trening"
elif significant and not b_more_accurate:
    recommendation = "Model A – znacząco lepsza accuracy"
elif not significant and a_faster:
    recommendation = "Model A – podobna accuracy przy krótszym czasie treningu"
else:
    recommendation = "Model B – podobna accuracy przy krótszym czasie treningu"

print(f"Rekomendacja: {recommendation}")
```
