```python
probabilities = [0.23, 0.67, 0.89, 0.34, 0.91, 0.12, 0.78, 0.45]
true_labels = [0, 1, 1, 0, 1, 0, 1, 0]
thresholds = [0.3, 0.5, 0.7]
```

```python
results = []

for threshold in thresholds:
    correct = 0
    for prob, true in zip(probabilities, true_labels):
        predicted = 1 if prob >= threshold else 0
        if predicted == true:
            correct += 1
    accuracy = correct / len(true_labels)
    results.append((threshold, accuracy))
    print(f"Threshold {threshold:.1f}: accuracy = {accuracy:.2%} ({correct}/{len(true_labels)})")
```

```python
best_threshold, best_accuracy = max(results, key=lambda item: item[1])
print(f"\nOptymalny threshold: {best_threshold:.1f} (accuracy = {best_accuracy:.2%})")
```
