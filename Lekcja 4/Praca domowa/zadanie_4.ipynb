```python
import pandas as pd
import re

def clean_column_name(name):
    # 1. Zastąpienie "_" spacjami
    name = name.replace("_", " ")
    
    # 2. Usunięcie nadmiarowych białych znaków (ze środka, początku i końca)
    name = re.sub(r'\s+', ' ', name).strip()
    
    # 3. Zamiana na małe litery, a następnie na Title Case
    # (Metoda .title() wewnętrznie traktuje tekst jako małe litery przed kapitalizacją)
    return name.title()

# --- Przykład użycia na DataFrame ---

# Tworzenie przykładowych danych z Twoją oryginalną nazwą kolumny
data = {
    " PATIENT age YEARS ": [25, 47, 31],   
}
df = pd.DataFrame(data)

print("Przed standaryzacją:")
print(df.columns.tolist())

# Zastosowanie transformacji na wszystkich kolumnach
df.columns = [clean_column_name(col) for col in df.columns]

print("\nPo standaryzacji:")
print(df.columns.tolist())
```
