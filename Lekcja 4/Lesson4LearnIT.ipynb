```python
# Zmienne można nadpisywać - zmienia się wtedy ich typ
liczba_probek = 150.5
# teraz to float, nie int
print(type(liczba_probek))
# <class 'float'>
```

```python
# Podstawowe operacje matematyczne na danych
liczba_treningowa = 120
liczba_testowa = 30
liczba_walidacyjna = 20
# Dodawanie
calkowita_liczba_probek = liczba_treningowa + liczba_testowa + liczba_walidacyjna
print(f"Całkowita liczba próbek: {calkowita_liczba_probek}")
# 170
# Odejmowanie
roznica = liczba_treningowa - liczba_testowa
print(f"Różnica między zbiorem treningowym a testowym: {roznica}")
# 90
# Mnożenie i dzielenie
podwojona_liczba = liczba_treningowa * 2
print(f"Podwojona liczba treningowa: {podwojona_liczba}")
# 240
# Dzielenie zwykłe (zawsze zwraca float)
udzial_treningowy = liczba_treningowa / calkowita_liczba_probek
print(f"Udział danych treningowych: {udzial_treningowy:.2f}")
# 0.71
# Dzielenie całkowite (zwraca int, odrzuca część ułamkową)
pelne_batche = calkowita_liczba_probek // 32
print(f"Liczba pełnych batch-ów (32 próbki): {pelne_batche}")
# 5
# Modulo - reszta z dzielenia (przydatne do batch processing)
pozostale_probki = calkowita_liczba_probek % 32
print(f"Pozostałe próbki po podziale na batch-e: {pozostale_probki}")
# 10
```

```python
# Potęgowanie
wariancja = 5.2
odchylenie_std = wariancja ** 0.5
# pierwiastek to potęga 0.5
print(f"Odchylenie standardowe: {odchylenie_std:.2f}")
# 2.28
```

```python
# Operatory porównania zwracają True lub False
accuracy_model_1 = 0.92
accuracy_model_2 = 0.88
prog_akceptacji = 0.90
# Porównania
print(f"Model 1 lepszy niż Model 2: {accuracy_model_1 > accuracy_model_2}")
# True
print(f"Model 1 osiągnął próg: {accuracy_model_1 >= prog_akceptacji}")# True
print(f"Model 2 poniżej progu: {accuracy_model_2 < prog_akceptacji}")# True
print(f"Modele równe: {accuracy_model_1 == accuracy_model_2}")# False
print(f"Modele różne: {accuracy_model_1 != accuracy_model_2}")# True
# Operatory logiczne: and, or, not
model1_ok = accuracy_model_1 >= prog_akceptacji
model2_ok = accuracy_model_2 >= prog_akceptacji
# and - oba warunki muszą być True
oba_modele_ok = model1_ok and model2_ok
print(f"Oba modele OK: {oba_modele_ok}")
# False
# or - przynajmniej jeden warunek True
przynajmniej_jeden_ok = model1_ok or model2_ok
print(f"Przynajmniej jeden model OK: {przynajmniej_jeden_ok}")
# True
# not - negacja
model2_wymaga_poprawy = not model2_ok
print(f"Model 2 wymaga poprawy: {model2_wymaga_poprawy}")
# Łączenie warunków - przykład z filtrowaniem danych
# True
wiek_pacjenta = 45
cisnienie = 140
# Pacjent wysokiego ryzyka: wiek > 40 AND ciśnienie > 130
wysokie_ryzyko = (wiek_pacjenta > 40) and (cisnienie > 130)
print(f"Pacjent wysokiego ryzyka: {wysokie_ryzyko}")
# True
```
