```python
import sys
import psutil
import platform
```

```python
# Informacje o systemie
print("=== INFORMACJE O SYSTEMIE ===")
print(f"System operacyjny: {platform.system()} {platform.release()}")
print(f"Python version: {sys.version}")
print()
```

```python
# Informacje o pamięci RAM
print("=== PAMIĘĆ RAM ===")
ram_gb = psutil.virtual_memory().total / (1024**3)
ram_available_gb = psutil.virtual_memory().available / (1024**3)
print(f"Całkowita RAM: {ram_gb:.2f} GB")
print(f"Dostępna RAM: {ram_available_gb:.2f} GB")
print()
```

```python
# Informacje o CPU
print("=== PROCESOR ===")
print(f"Liczba rdzeni CPU: {psutil.cpu_count()}")
print(f"Obciążenie CPU: {psutil.cpu_percent(interval=1)}%")
print()
```

```python
# Sprawdzenie dostępności GPU (w Colab)
try:
  import torch
  if torch.cuda.is_available():
      print("=== GPU ===")
      print(f"GPU dostępne: Tak")
      print(f"Nazwa GPU: {torch.cuda.get_device_name(0)}")
      print(f"Pamięć GPU: {torch.cuda.get_device_properties(0).total_memory / (1024**3):.2f} GB")
  else:
    print("GPU: Niedostępne (włącz GPU w Runtime -> Change runtime type)")
except ImportError:
  print("PyTorch nie zainstalowany - nie można sprawdzić GPU")
```

```python
# Przykład 2
# Montowanie Google Drive do Colab
from google.colab import drive
import os
# Montujemy dysk (pojawi się prośba o autoryzację)
drive.mount('/content/drive')
# Sprawdzamy, czy dysk został zamontowany
if os.path.exists('/content/drive/MyDrive'):
  print("✓ Google Drive pomyślnie zamontowany!")
  # Tworzymy folder dla naszego projektu Data Science
  project_folder = '/content/drive/MyDrive/DataScience_Projects'
  os.makedirs(project_folder, exist_ok=True)
  print(f"✓ Folder projektu: {project_folder}")
  # Wyświetlamy zawartość folderu
  print("\nZawartość folderu MyDrive:")
  for item in os.listdir('/content/drive/MyDrive')[:10]:
    print(f" # Pierwsze 10 elementów - {item}")
else:
  print("✗ Problem z montowaniem Google Drive")
```
