# Przykładowy notebook do testu monitora

Stopniowo zajmuje pamięć i liczy mnożenia macierzy, żeby było widać zmiany na wykresach.

```python
import time
import numpy as np

np.random.seed(42)
```

```python
# zajmujemy RAM w porcjach po ~400 MB
data = []
for i in range(8):
    data.append(np.random.rand(50_000_000))
    time.sleep(0.5)
```

```python
# obciążenie CPU - mnożenie macierzy
a = np.random.rand(3000, 3000)
for i in range(10):
    b = a @ a
print(b.shape)
```

```python
# zwalniamy pamięć
del data
time.sleep(2)
```
