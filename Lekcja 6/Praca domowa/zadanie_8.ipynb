```python
sprzedaz = {
    (2024, 'Q1'): 120_000,
    (2024, 'Q2'): 140_000,
    (2024, 'Q3'): 135_000,
    (2024, 'Q4'): 160_000
}

print(sprzedaz[(2024, 'Q2')])

sprzedaz[(2025, 'Q1')] = 130_000

print(f"Najwyższa: {max(sprzedaz.values())}")
print(f"Najniższa: {min(sprzedaz.values())}")
print(f"Łącznie: {sum(sprzedaz.values())}")
print(f"Liczba kwartałów: {len(sprzedaz)}")
```
